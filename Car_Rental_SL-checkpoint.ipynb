### Importing all the required python libraries

```python
from datetime import datetime, timedelta
import math
```

### Settng Price according to the time

```python
HOURLY_RATE = 10     # dollars per car per hour
DAILY_RATE = 50      # dollars per car per day
WEEKLY_RATE = 200    # dollars per car per week
```

## CarRental Class

The CarRental class manages all car-rental operations:

- Tracks how many cars are available

- Lets customers rent cars (hourly, daily, weekly)

- Records rental time

- Calculates bills when cars are returned

```python
# ---- CarRental class ----
class CarRental:
    """Minimal car rental manager: tracks stock, rent, and return."""

    def __init__(self, total_cars):
        # why: set initial inventory
        self.stock = int(total_cars)

    def display_available(self):
        """Return available car count."""
        return self.stock

    def rent(self, num_cars, mode):
        """
        Rent num_cars using mode: 'hourly', 'daily', or 'weekly'.
        Returns a rental record dict or raises ValueError.
        """
        # Validate
        if num_cars <= 0:
            raise ValueError("Must rent at least 1 car.")
        if num_cars > self.stock:
            raise ValueError("Not enough cars available.")

        # Record rental
        rental_time = datetime.now()
        self.stock -= num_cars
        return {"rental_time": rental_time, "mode": mode, "num": num_cars}

    def process_return(self, record):
        """
        Accept a rental record, compute bill, update stock, and return bill dict.
        """
        if not record:
            raise ValueError("No rental record provided.")

        start = record["rental_time"]
        mode = record["mode"]
        num = record["num"]

        end = datetime.now()
        delta = end - start
        seconds = delta.total_seconds()

        # compute billed units (round up)
        if mode == "hourly":
            units = math.ceil(seconds / 3600)
            rate = HOURLY_RATE
        elif mode == "daily":
            units = math.ceil(seconds / (3600*24))
            rate = DAILY_RATE
        else:  # weekly
            units = math.ceil(seconds / (3600*24*7))
            rate = WEEKLY_RATE

        amount = units * rate * num

        # Put cars back
        self.stock += num

        bill = {
            "mode": mode,
            "start": start,
            "end": end,
            "units": units,
            "rate": rate,
            "num_cars": num,
            "amount": round(amount, 2),
            "stock_after": self.stock
        }
        return bill
```

## Customer Class

The Customer class stores information about a customer’s current rental.

 - It does not process payments.

 - Holds the active rental record

 - Gives the record back when the customer returns cars

```python

# ---- Customer class ----
class Customer:
    """Stores a single active rental for simplicity."""
    def __init__(self):
        self.rental = None

    def take_rental(self, record):
        self.rental = record

    def give_back(self):
        r = self.rental
        self.rental = None
        return r

# ---- Simple text menu ----
def main():
    shop = CarRental(total_cars=5)   # start with 5 cars (changeable)
    user = Customer()

    while True:
        print("\n--- Simple Car Rental ---")
        print("1) Show available cars")
        print("2) Rent hourly")
        print("3) Rent daily")
        print("4) Rent weekly")
        print("5) Return car(s)")
        print("6) Exit")
        choice = input("Choose (1-6): ").strip()

        if choice == "1":
            print("Available cars:", shop.display_available())

        elif choice in ("2", "3", "4"):
            try:
                n = int(input("How many cars to rent? ").strip())
            except ValueError:
                print("Enter a number.")
                continue

            mode = {"2": "hourly", "3": "daily", "4": "weekly"}[choice]
            try:
                record = shop.rent(n, mode)
            except Exception as e:
                print("Error:", e)
                continue

            user.take_rental(record)
            print(f"Rented {n} car(s) on {mode} basis at {record['rental_time']}")
            print("Cars left:", shop.display_available())

        elif choice == "5":
            record = user.give_back()
            if not record:
                print("You have no active rental.")
                continue
            bill = shop.process_return(record)
            print("\n--- Bill ---")
            print("Mode:", bill["mode"])
            print("Start:", bill["start"])
            print("End:  ", bill["end"])
            print("Units billed:", bill["units"])
            print("Rate per unit:", bill["rate"])
            print("Number of cars:", bill["num_cars"])
            print("Total amount: $", bill["amount"])
            print("Available now:", bill["stock_after"])

        elif choice == "6":
            print("Bye!")
            break
        else:
            print("Invalid choice. Type 1-6.")

```
